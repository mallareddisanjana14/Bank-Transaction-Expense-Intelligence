```python
customer_risk = df_clean.groupby("customer_id").agg(
    transaction_count=("transaction_id", "count"),
    total_spending=("transaction_amount", "sum"),
    fraud_count=("is_fraud", "sum"),
    average_transaction=("transaction_amount", "mean"),
    max_transaction=("transaction_amount", "max")
)

customer_risk["risk_score"] = (
    (customer_risk["fraud_count"] > 0).astype(int) * 3 +
    (customer_risk["max_transaction"] > 100000).astype(int) * 1 +
    (customer_risk["transaction_count"] > 100).astype(int) * 1
)

customer_risk["risk_level"] = pd.cut(
    customer_risk["risk_score"],
    bins=[-1, 1, 3, 5],
    labels=["Low", "Medium", "High"]
)

display(
    customer_risk.sort_values(
        "risk_score",
        ascending=False
    ).head(20)
)
```
